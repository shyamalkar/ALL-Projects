```python
import pandas as pd 
import numpy as np 
import matplotlib.pyplot as plt
import seaborn as sns 
import plotly as pl

```

# Load and Understand the dataset

```python
data_load = pd.read_csv("/Users/shyamalkar/Desktop/DS_project/Dataset_only/Expanded_Indian_Travel_Dataset.csv")
df = data_load
df.head()
```

# Let's start the EDA section 


# State vs Popular Attraction 

```python
plt.figure(figsize=(14, 7))

sns.countplot(data=df,x="State",hue="Popular Attraction")

plt.title("Popular Attractions by State")
plt.xlabel("State")
plt.ylabel("Number of Destinations")

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.countplot(data=df,x="Region",order=df["Region"].value_counts().index)

plt.title("Tourist Destinations by Region")
plt.xlabel("Region")
plt.ylabel("Number of Destinations")

plt.tight_layout()
plt.show()
```

# Category distribution 

```python
plt.figure(figsize=(10, 6))

sns.countplot(
    data=df,
    y="Category",
    order=df["Category"].value_counts().index
)

plt.title("Tourist Destinations by Category")
plt.xlabel("Number of Destinations")
plt.ylabel("Category")

plt.tight_layout()
plt.show()
```

# Accessibility and distribution 

```python
plt.figure(figsize=(8, 5))

sns.countplot(
    data=df,
    x="Accessibility",
    order=df["Accessibility"].value_counts().index
)

plt.title("Tourist Destination Accessibility")
plt.xlabel("Accessibility")
plt.ylabel("Number of Destinations")

plt.tight_layout()
plt.show()
```

```python
region_category = pd.crosstab(
    df["Region"],
    df["Category"]
)

plt.figure(figsize=(12, 7))

sns.heatmap(
    region_category,
    annot=True,
    fmt="d"
)

plt.title("Tourist Category Distribution by Region")
plt.xlabel("Category")
plt.ylabel("Region")

plt.tight_layout()
plt.show()
```

```python
state_category = pd.crosstab(
    df["State"],
    df["Category"]
)

plt.figure(figsize=(14, 10))

sns.heatmap(
    state_category,
    annot=True,
    fmt="d"
)

plt.title("Tourist Category Distribution by State")
plt.xlabel("Category")
plt.ylabel("State")

plt.tight_layout()
plt.show()
```

```python
top_railway = (
    df["Nearest Railway Station"]
    .value_counts()
    .head(15)
    .index
)

plt.figure(figsize=(12, 7))

sns.countplot(
    data=df[
        df["Nearest Railway Station"].isin(top_railway)
    ],
    y="Nearest Railway Station",
    order=top_railway
)

plt.title("Top 15 Nearest Railway Stations")
plt.xlabel("Number of Destinations")
plt.ylabel("Nearest Railway Station")

plt.tight_layout()
plt.show()
```
