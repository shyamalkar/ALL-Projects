# Data Visualization with EDA

`EDA help us to see the raw data into a suitable visualization form`

```python
plt.figure(figsize=(7, 5))

sns.countplot(
    data=df,
    x="readmitted"
)

plt.xlabel("Readmission")
plt.ylabel("Number of Patients")
plt.title("Distribution of Patient Readmission")

plt.show()
```

# Numerical feature distribution

```python
numeric_cols = [
    "time_in_hospital",
    "num_lab_procedures",
    "num_procedures",
    "num_medications",
    "number_outpatient",
    "number_emergency",
    "number_inpatient",
    "number_diagnoses"
]
```

```python
# Create figure
fig, axes = plt.subplots(
    4, 2,
    figsize=(14, 18)
)

# Flatten axes
axes = axes.flatten()

# Plot each numerical feature
for i, column in enumerate(numeric_cols):

    sns.histplot(
        data=df,
        x=column,
        kde=True,
        ax=axes[i]
    )

    axes[i].set_title(f"Distribution of {column}")

plt.tight_layout()
plt.show()
```

```python
# Box plot for outlier detection
fig, axes = plt.subplots(
    4, 2,
    figsize=(14, 18)
)

axes = axes.flatten()

for i, column in enumerate(numeric_cols):

    sns.boxplot(
        data=df,
        x=column,
        ax=axes[i]
    )

    axes[i].set_title(f"Outliers in {column}")

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 6))

sns.boxplot(
    data=df,
    x="readmitted",
    y="time_in_hospital"
)

plt.xlabel("Readmission")
plt.ylabel("Time in Hospital")

plt.title("Hospital Stay vs Readmission")

plt.show()
```
