# Let's Visualize Using EDA 

```python
plt.figure(figsize=(7, 5)) # set size of the visualizer . 

sns.countplot(data=df,x="diagnosis")

plt.title("Diagnosis Distribution")
plt.xlabel("Diagnosis")
plt.ylabel("Number of Samples")

plt.show()
```

# Radius mean vs diagnosis

`This is one of the first relationships i would investigate.`

```python
plt.figure(figsize=(8, 5))

sns.boxplot(
    data=df,
    x="diagnosis",
    y="radius_mean"
)

plt.title("Radius Mean Distribution by Diagnosis")
plt.xlabel("Diagnosis")
plt.ylabel("Radius Mean")

plt.show()
```

# Area mean vs diagnosis 

```python
plt.figure(figsize=(8, 5))

sns.boxplot(
    data=df,
    x="diagnosis",
    y="area_mean"
)

plt.title("Area Mean Distribution by Diagnosis")
plt.xlabel("Diagnosis")
plt.ylabel("Area Mean")

plt.show()
```

# Concavity vs diagnosis 

`This is another important feature group.`

```python
plt.figure(figsize=(8, 5))

sns.boxplot(
    data=df,
    x="diagnosis",
    y="concavity_mean"
)

plt.title("Concavity Mean by Diagnosis")
plt.xlabel("Diagnosis")
plt.ylabel("Concavity Mean")

plt.show()
```

# This breast cancer dataset deserve feature engineering  but small amout 

# Mean -> wrost difference / ratio 

```python
df["radius_change"] = df["radius_worst"] - df["radius_mean"] #actual value of radius_change 0 position value and radius_mean 0 position value like :- 17.99 - 25.38 = 7.39


df["perimeter_change"] = df["perimeter_worst"] - df["perimeter_mean"]

df["area_change"] = df["area_worst"] - df["area_mean"]
```
