# EDA (Exploratory Data Analysis)

`Let's Findout what inside in this dataset`

```python
df.columns
```

# Male and Female 

```python
plt.figure(figsize=(7, 5))

sns.countplot(
    data=df,
    x="Gender"
)

plt.title("Gender Distribution")
plt.show()
```

# Education level 

```python
plt.figure(figsize=(9, 5))

sns.countplot(
    data=df,
    x="Education_Level"
)

plt.title("Education Level Distribution")
plt.xticks(rotation=45)

plt.show()
```

# Male and Female Pie charts. 

```python
gender_counts = df["Gender"].value_counts()

plt.figure(figsize=(7, 7))

plt.pie(
    gender_counts.values,
    labels=gender_counts.index,
    autopct="%1.1f%%"
)

plt.title("Gender Distribution")

plt.show()
```

# Age analysis 

```python
plt.figure(figsize=(8, 5))

sns.histplot(
    data=df,
    x="Age",
    bins=20,
    kde=True
)

plt.title("Age Distribution")
plt.show()
```

# Age Box-plot

```python
plt.figure(figsize=(8, 4))

sns.boxplot(
    data=df,
    x="Age"
)

plt.title("Age Boxplot")
plt.show()
```

# Social media vs mental health 

```python
plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x="Daily_Social_Media_Hours",
    y="Mental_Health_Score"
)

plt.title("Social Media Usage vs Mental Health")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.boxplot(
    data=df,
    x="Gender",
    y="Mental_Health_Score"
)

plt.title("Gender vs Mental Health Score")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.boxplot(
    data=df,
    x="Gender",
    y="Physical_Health_Score"
)

plt.title("Gender vs Physical Health Score")

plt.show()
```

```python
plt.figure(figsize=(10, 5))

sns.boxplot(
    data=df,
    x="Education_Level",
    y="Mental_Health_Score"
)

plt.title("Education Level vs Mental Health")

plt.xticks(rotation=45)

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.regplot(
    data=df,
    x="Sleep_Hours",
    y="Mental_Health_Score"
)

plt.title("Sleep Hours vs Mental Health")

plt.show()
```
