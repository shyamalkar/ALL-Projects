# Prepare the Features and target 
`My target is Loan status `

# Separate X and y 

```python
# 1. Drop rows where Loan_Status is missing BEFORE separating X and y
df = df.dropna(subset=["Loan_Status"])

# 2. X contains all input features
X = df.drop("Loan_Status", axis=1)

# 3. y contains only the target column, cleaned of spaces, and mapped safely
y = df["Loan_Status"].astype(str).str.strip().map({'N': 0, 'Y': 1})
```

```python
dfw = df['Loan_Status'].unique()
print(dfw)
```

# Split into training and testing data

```python
df.isnull().sum()
```

```python
# Split data into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(
    X,                         # Input features
    y,                         # Target
    test_size=0.20,            # 20% → testing data
    random_state=42,           # Reproducible split
    stratify=y                 # Preserve Y/N proportion
)
```

```python
# Check the shape of each dataset
print("X_train:", X_train.shape)
print("X_test :", X_test.shape)

print("y_train:", y_train.shape)
print("y_test :", y_test.shape)
```

```python
df.dtypes
```

# Categorical and numerical columns

```python
# Categorical columns → need One-Hot Encoding
categorical_features = [
    "Gender",
    "Married",
    "Education",
    "Self_Employed",
    "Property_Area"
]

# Numerical columns → don't need encoding
numeric_features = [
    "Dependents",
    "ApplicantIncome",
    "CoapplicantIncome",
    "LoanAmount",
    "Loan_Amount_Term",
    "Credit_History",
    "TotalIncome"
]
```

```python
df.dtypes
```

# Numerical Preprocessing 

```python
numeric_pipeline = Pipeline([

    # Fill missing numerical values using median
    ("imputer", SimpleImputer(strategy="median")),

    # Scale numerical features
    ("scaler", StandardScaler())
])
```

# Categorical Preprocessing 

```python
categorical_pipeline = Pipeline([

    # Fill missing categorical values
    # using the most frequent category
    ("imputer", SimpleImputer(strategy="most_frequent")),

    # ⭐ ENCODING HAPPENS HERE ⭐
    # Convert categorical values into numerical columns
    ("encoder", OneHotEncoder(
        handle_unknown="ignore"
    ))
])
```

# Combine numerical + Categorical proprocessing 

```python
preprocessor = ColumnTransformer([

    # Numerical columns
    # → Imputation → Scaling
    ("num", numeric_pipeline, numeric_features),

    # Categorical columns
    # → Imputation → One-Hot Encoding
    ("cat", categorical_pipeline, categorical_features)
])
```

# Create XGBoost model 

```python
xgb_model = XGBClassifier(

    # Number of boosting trees
    n_estimators=200,

    # Contribution of each tree
    learning_rate=0.05,

    # Maximum depth of each tree
    max_depth=3,

    # Use 80% of training rows for each tree
    subsample=0.8,

    # Use 80% of features for each tree
    colsample_bytree=0.8,

    # Binary classification
    objective="binary:logistic",

    # Loss metric
    eval_metric="logloss",

    # Reproducibility
    random_state=42,

    # Use all available CPU cores
    n_jobs=-1
)

```

# XGBoost Pipeline 

```python
xgb_pipeline = Pipeline([

    # Step 1:
    # Imputation + Encoding + Scaling
    ("preprocessor", preprocessor),

    # Step 2:
    # XGBoost
    ("model", xgb_model)
])
```

```python
xgb_pipeline.fit(
    X_train,
    y_train
)
```

# Make Predictions 

```python
y_pred = xgb_pipeline.predict(X_test)
```

# Get Prediction probabilities 

```python
# Probability of class 1 = Approved
y_prob = xgb_pipeline.predict_proba(X_test)[:, 1]
```

# Evaluate Model 

```python
from xgboost import XGBClassifier

# Evaluation
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    confusion_matrix,
    classification_report
)
```

```python
accuracy = accuracy_score(
    y_test,
    y_pred
)

precision = precision_score(
    y_test,
    y_pred
)

recall = recall_score(
    y_test,
    y_pred
)

f1 = f1_score(
    y_test,
    y_pred
)

roc_auc = roc_auc_score(
    y_test,
    y_prob
)

```

# Display the results

```python
print("=" * 45)
print("           XGBOOST RESULTS")
print("=" * 45)

print(f"Accuracy  : {accuracy:.4f}")
print(f"Precision : {precision:.4f}")
print(f"Recall    : {recall:.4f}")
print(f"F1 Score  : {f1:.4f}")
print(f"ROC-AUC   : {roc_auc:.4f}")
```

# Hyperparameter tuning 

```python
from sklearn.model_selection import RandomizedSearchCV
param_grid = {

    # Number of trees
    "model__n_estimators": [
        100,
        200,
        300,
        500
    ],

    # Learning rate
    "model__learning_rate": [
        0.01,
        0.03,
        0.05,
        0.1
    ],

    # Tree depth
    "model__max_depth": [
        2,
        3,
        4,
        5
    ],

    # Minimum child weight
    "model__min_child_weight": [
        1,
        3,
        5
    ],

    # Row sampling
    "model__subsample": [
        0.7,
        0.8,
        0.9,
        1.0
    ],

    # Feature sampling
    "model__colsample_bytree": [
        0.7,
        0.8,
        0.9,
        1.0
    ],

    # Minimum loss reduction
    "model__gamma": [
        0,
        0.1,
        0.3,
        0.5
    ],

    # L1 regularization
    "model__reg_alpha": [
        0,
        0.01,
        0.1,
        1
    ],

    # L2 regularization
    "model__reg_lambda": [
        1,
        2,
        5
    ]
}
```

# Randomized search 

```python
random_search = RandomizedSearchCV(

    # Complete preprocessing + XGBoost pipeline
    estimator=xgb_pipeline,

    # Hyperparameter combinations
    param_distributions=param_grid,

    # Number of combinations to test
    n_iter=30,

    # 5-fold cross-validation
    cv=5,

    # Optimize F1 Score
    scoring="f1",

    # Use all CPU cores
    n_jobs=-1,

    # Reproducibility
    random_state=42,

    # Show progress
    verbose=1
)
```

# Train Randomized search 

```python
random_search.fit(
    X_train,
    y_train
)
```

# Best parameters 

```python
print("\nBest Parameters:")
print(random_search.best_params_)

print("\nBest Cross-Validation F1 Score:")
print(random_search.best_score_)
```

# Feature importance 

Because you used a Pipeline + OneHotEncoder, we need to correctly extract the feature names and XGBoost importance.

```python
# Get the trained XGBoost model from the pipeline
xgb_final = final_model.named_steps["model"]

# Get the fitted preprocessor
fitted_preprocessor = final_model.named_steps["preprocessor"]

# Get feature names after encoding
feature_names = fitted_preprocessor.get_feature_names_out()

# Get XGBoost feature importance
importance = xgb_final.feature_importances_

# Create a DataFrame
feature_importance_df = pd.DataFrame({
    "Feature": feature_names,
    "Importance": importance
})

# Sort from highest to lowest importance
feature_importance_df = feature_importance_df.sort_values(
    by="Importance",
    ascending=False
)

print(feature_importance_df.head(15))
```

```python
plt.figure(figsize=(10, 6))

sns.barplot(
    data=feature_importance_df.head(15),
    x="Importance",
    y="Feature"
)

plt.title("Top 15 XGBoost Feature Importance")
plt.xlabel("Importance")
plt.ylabel("Feature")

plt.show()
```
